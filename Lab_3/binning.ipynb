# Binning white noise

For uncorrelated (white) noise, binning $b$ consecutive points should reduce the r.m.s. as $1/\sqrt{b}$. Here we check that numerically.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

# Generate some fake white-noise data
data = rng.normal(0.0, 1.0, 1000)
```

```python
bmax = 100  # maximum bin size to try
binsize = np.arange(2, bmax + 1)
rms = np.empty(binsize.size)

for k, b in enumerate(binsize):
    # Bin the time series: average each block of b consecutive points
    imax = len(data) // b                          # number of complete bins
    binned = data[:imax * b].reshape(imax, b).mean(axis=1)
    rms[k] = np.std(binned)

# Theoretical r.m.s., with a small-sample correction for the number of bins
imax = len(data) // binsize
rmstheory = binsize**-0.5 * np.sqrt(imax / (imax - 1))
```

```python
fig, ax = plt.subplots()
ax.scatter(binsize, rms, color="k", label="binned data")
ax.plot(binsize, rmstheory, color="r", label=r"$1/\sqrt{b}$ (theory)")
ax.set_xlabel("bin size")
ax.set_ylabel("rms")
ax.legend()
plt.show()
```
